#### Demo: With TF-IDF 

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "Employees can request time off through the HR portal at least two weeks in advance.",
    "The company laptop replacement cycle is every three years for engineering staff.",
    "To expense a business trip, submit receipts within 30 days via the finance system.",
    "New hires must complete security training within their first week of employment.",
    "Remote employees are eligible for a one-time $500 home office stipend.",
]
query = "How do I get a new computer for work?"

vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(docs)
query_vector = vectorizer.transform([query])

scores = cosine_similarity(query_vector, doc_vectors)[0]

ranked = sorted(zip(scores, docs), reverse=True)
print(f"Query: {query}\n")
for score, doc in ranked:
    print(f"{score:.3f}  {doc}")
```

#### Observations:

The correct answer (Doc 2, laptops) scores:

-   TF-IDF has low signal for it, because "computer," "get," "new," and "work" don't appear in that document at all.
-   The top-ranked result is only weakly related, matched on the generic overlap of "work"/"employment." 
-   This is the failure mode to remember.

#### Demo: With embeddings

```python
from sentence_transformers import SentenceTransformer, util

docs = [
    "Employees can request time off through the HR portal at least two weeks in advance.",
    "The company laptop replacement cycle is every three years for engineering staff.",
    "To expense a business trip, submit receipts within 30 days via the finance system.",
    "New hires must complete security training within their first week of employment.",
    "Remote employees are eligible for a one-time $500 home office stipend.",
]
query = "How do I get a new computer for work?"

model = SentenceTransformer("all-MiniLM-L6-v2")

doc_embeddings = model.encode(docs, convert_to_tensor=True)
query_embedding = model.encode(query, convert_to_tensor=True)

scores = util.cos_sim(query_embedding, doc_embeddings)[0]

ranked = sorted(zip(scores.tolist(), docs), reverse=True)
print(f"Query: {query}\n")
for score, doc in ranked:
    print(f"{score:.3f}  {doc}")
```

#### Observations

Doc 2 jumps from (compared to TF-IDF) to the clear top match, because 

- "computer" and "laptop" sit close together in embedding space, 
- and "get a new [thing] for work" aligns semantically with "replacement cycle."

#### Demo: Full RAG answer generation

```python
from sentence_transformers import SentenceTransformer, util
import ollama

docs = [
    "Employees can request time off through the HR portal at least two weeks in advance.",
    "The company laptop replacement cycle is every three years for engineering staff.",
    "To expense a business trip, submit receipts within 30 days via the finance system.",
    "New hires must complete security training within their first week of employment.",
    "Remote employees are eligible for a one-time $500 home office stipend.",
]
query = "How do I get a new computer for work?"

# --- Retrieval (embeddings, from Slide 38) ---
model = SentenceTransformer("all-MiniLM-L6-v2")
doc_embeddings = model.encode(docs, convert_to_tensor=True)
query_embedding = model.encode(query, convert_to_tensor=True)
scores = util.cos_sim(query_embedding, doc_embeddings)[0]
best_idx = scores.argmax().item()
retrieved_chunk = docs[best_idx]

print(f"Retrieved chunk: {retrieved_chunk}\n")

# --- Augmentation ---
prompt = f"""You are an internal HR/IT assistant. Answer the employee's question
using ONLY the information in the context below. If the context doesn't contain
the answer, say you don't have that information. Cite the policy in your answer.

Context:
{retrieved_chunk}

Question: {query}

Answer:"""

# --- Generation (local LLM via Ollama) ---
response = ollama.generate(model="llama3.2", prompt=prompt)
print("Generated answer:")
print(response["response"])
```

pypdf is fine for straightforward text-based PDFs (most policy docs, reports, Word-exported PDFs). Two things worth flagging to students:

Scanned PDFs (images of text) return empty or garbled text — pypdf doesn't do OCR. If you upload a scanned document, you'd need pytesseract + pdf2image for OCR first. Worth a quick "does your PDF have selectable text?" sanity check before debugging further.

Tables and multi-column layouts often extract messily — text can interleave columns or lose table structure. pdfplumber handles tables and layout better than pypdf if that comes up.


Chunking a PDF is a bigger deal than chunking short strings

This is the part actually worth spending time on  — 

-   your corpus so far was 5 clean one-liners; 
-   a real PDF is a wall of text, so chunking strategy suddenly matters in practice, not just in theory.

```python
# This is naive fixed-size chunking (character count, with overlap)
# Explore more sophisticated chunking methods for better results.

def chunk_text(text, chunk_size=500, overlap=50):
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start += chunk_size - overlap
    return chunks

pdf_text = extract_text_from_pdf("employee_handbook.pdf")
docs = chunk_text(pdf_text)

```
