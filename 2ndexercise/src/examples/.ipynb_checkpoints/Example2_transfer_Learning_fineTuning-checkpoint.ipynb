
# Transfer Learning by Finetuning a ConvNet

Rememeber that fine-tuning means that,
- We shall take a pretrained model and use its pretrained weights (its weights when it was pretrained and released for us) as the starting point to train the model on our custom dataset.
- Therefore we do not freeze any layers and when the model is trained, it would have adjusted its weights to predict on the new dataset.
- We still did not have to train it from scratch.

Now, we could still have used feature extraction for this application, but the purpose here was to show how to setup for fine-tuning a model.

We will also learn a few more things about using PyTorch commands available for our convenience.


```python
# Import modules

import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim import lr_scheduler
import torchvision
from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader

from torchinfo import summary

import numpy as np
import matplotlib.pyplot as plt
import time
import os
from PIL import Image
```

## Load Data

The problem we're going to solve today is to train a model to classify
**ants** and **bees**. We have about 120 training images each for ants and bees.
There are 75 validation images for each class. Usually, this is a very
small dataset to generalize upon, if trained from scratch. Since we
are using transfer learning, we should be able to generalize reasonably
well.


Note ::
   Download the data from
   [here](https://download.pytorch.org/tutorial/hymenoptera_data.zip)
   and extract it to the current directory. You will have a directory called 'hymenoptera_data' and with that the train and val directories.



```python
# We will use the pretrained model called "resnet18" another CNN model and so lets get the weights
weights = models.ResNet18_Weights.DEFAULT # .DEFAULT = best available weights from pretraining
print(f"The weights are: {weights}")

# Get transforms the model was trained with
auto_transforms =  weights.transforms()
print(f"The tranforms used in this model are:\n{auto_transforms}")
```

```python
# Data augmentation with data_transforms and normalization for training data 
# We will use the auto_transforms for the test data as we do not need to do any data augmentation
data_transforms = transforms.Compose([
        transforms.RandomResizedCrop(224),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
    ])

data_dir = './hymenoptera_data' 

# Load data
train_data = datasets.ImageFolder(data_dir + '/train', transform=data_transforms)
test_data = datasets.ImageFolder(data_dir + '/val', transform=auto_transforms)

# Setup dataloaders
train_loader = DataLoader(train_data,batch_size=4,shuffle=True)
test_loader = DataLoader(test_data,batch_size=4)

# Get the classes
classes = train_data.classes
print(f"The classes are : {classes}")


# Choose to use GPU or CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

### Visualize a few images
Let's visualize a few training images so as to understand the data
augmentations.



```python

def imshow(inp, title=None):
  """Display image for Tensor."""
  inp = inp.numpy().transpose((1, 2, 0)) 
  mean = np.array([0.485, 0.456, 0.406])
  std = np.array([0.229, 0.224, 0.225])
  inp = std * inp + mean

  inp = np.clip(inp, 0, 1)
  plt.imshow(inp)
  if title is not None:
    plt.title(title)
  plt.pause(0.001)


# Get a batch of training data
inputs, labels = next(iter(train_loader))

# Make a grid from batch
out = torchvision.utils.make_grid(inputs)

imshow(out, title=[classes[x]for x in labels])
```

## Training the model

Now, let's write a general function to train a model. 


```python
def train_model(model, criterion, optimizer, num_epochs=25):
    since = time.time()

    best_acc = 0 # Corrected by moving out of the for loop

    for epoch in range(num_epochs):
        print(f'Epoch {epoch}/{num_epochs - 1}')
        print('-' * 10)

        # Put model in training mode
        model.train()

        running_loss = 0.0
        running_corrects = 0
        

        # Iterate over data.
        for inputs, labels in train_loader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            # zero the parameter gradients
            optimizer.zero_grad()

            # forward pass
            y_pred = model(inputs)
            
            loss = criterion(y_pred, labels)

            # back propogate
            loss.backward()
            optimizer.step()

            # statistics
            running_loss += loss.item()

            preds = torch.argmax(torch.softmax(y_pred, dim=1), dim=1)   
            running_corrects += ((preds == labels).sum().item()/len(y_pred))

            # Test loop
           # Put model in eval mode
            model.eval() 
        
            # Setup test loss and test accuracy values
            test_loss, test_acc = 0, 0
        
            # Turn off gradient backpropogation while in inference mode
            with torch.no_grad():
                # Loop through DataLoader batches
                for batch, (X, y) in enumerate(test_loader):
                    # Send data to target device
                    X, y = X.to(device), y.to(device)
        
                    # 1. Forward pass
                    test_pred_logits = model(X)
        
                    # 2. Calculate and accumulate loss
                    loss = criterion(test_pred_logits, y)
                    test_loss += loss.item()
        
                    # Calculate and accumulate accuracy
                    test_pred_labels = test_pred_logits.argmax(dim=1)
                    test_acc += ((test_pred_labels == y).sum().item()/len(test_pred_labels))

        train_loss = running_loss / len(train_loader)
        train_acc = running_corrects / len(train_loader)
        test_loss = test_loss / len(test_loader)
        test_acc = test_acc / len(test_loader)

        print(f"Epoch:{epoch} Average Train Loss: {train_loss:.3f} Average Train acc: {train_acc:.2f} Average Test Loss: {test_loss:.3f} Average Test Acc: {test_acc:.2f}")

        # deep copy the model
        if test_acc > best_acc:
            best_acc = test_acc
            torch.save(model.state_dict(), 'best_model_params.pt')

    print()

    time_elapsed = time.time() - since
    print(f'Training complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Best val Acc: {best_acc:4f}')

    # load best model weights
    model.load_state_dict(torch.load('best_model_params.pt'))
    return model
```

## Finetuning the ConvNet (CNN)

Load a pretrained model and reset final fully connected layer.




```python
# These pretrained models are all different in how they are structured and we always have to print the model to understand what the layers are called
# or google search to get the exact names

model_ft = models.resnet18(weights=weights) # load the model for fine tuning with the 
num_ftrs = model_ft.fc.in_features  # resnet18 final fully connected layer (fc)

# Here the size of each output sample is set to 2.
# Alternatively, it can be generalized to ``nn.Linear(num_ftrs, len(class_names))``.
model_ft.fc = nn.Linear(num_ftrs, 2)   # We need to modify the final output of the model to 2

model_ft = model_ft.to(device)

# Define the loss function
criterion = nn.CrossEntropyLoss()

# Observe that all parameters are being optimized
optimizer = optim.Adam(model_ft.parameters(), lr=0.00001)
```

```python
# Let us print the model summary and we will see that all the layers are trainable
model_summary = summary(model_ft,(1,3,224,224), col_names=['output_size', 'kernel_size','num_params','trainable'])

print(f"Model Summary:\n{model_summary}")
```

### Train and evaluate

It should take around 15-25 min on CPU. On GPU though, it takes a few minutes.




```python
# The summary tells us that the model has 11.177 trainable parameters

model_ft = train_model(model_ft, criterion, optimizer, num_epochs=10)
```
